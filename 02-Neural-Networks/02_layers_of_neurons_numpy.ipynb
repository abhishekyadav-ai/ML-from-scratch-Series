Coding multiple neurons and stacking them up together using Numpy


Note: 
That we are avoiding the involvement of  activation function for the sake of simplicity

```python
#Importing packages
import numpy as np
```

```python
#Coding a single neuron using numpy
inputs=[1.0,2.0,3.0,2.5]
weights=[0.2,0.8,-0.5,1.0]

bias=[1,2,3,5]

outputs=np.dot(weights, inputs)+bias
print(outputs)
```

```python
#Coding layers of neurons using numpy
inputs=[1.0,2.0,3.0,2.5]

weights=[[0.2,0.8,-0.5,1],
         [0.5,-0.91,0.26,-0.5],
         [-0.26,-0.27,0.17,0.87]]
biases=[2.0,3.0,0.5]

layer_outputs=np.dot(weights, inputs)+biases
print(layer_outputs) 
```

```python
weights=np.array(weights)
```

```python
print(np.dot(inputs,weights.T)+biases)
```

```python
#Layers of Neurons and Batch of Data using Numpy

inputs= np.array([
    [1,2,3,2.5],
    [2,5,-1,2.0],
    [-1.5,2.7,3.3,0.8]
])


weights=np.array([[0.2,0.8,-0.5,1],
   [0.5,-0.91,0.26,-0.5],
   [-0.26,-0.27,0.17,0.87]
   ])
biases=np.array([2.0,3.0,0.5])

outputs=np.dot(inputs,weights.T)+biases
print(outputs)
```

Adding more layers


```python
inputs=[[1,2,3,2.5],
        [2.,5.,-1.,2],
        [-1.5,2.7,3.3,-0.8]]
weights=[[0.2,0.8,-0.5,1],
         [0.5,-0.91,0.26,-0.5],
         [-0.26,-0.27,0.17,0.87]]

biases=[2,3,0.5]

weights2=[[0.1,-0.14,0.5],
          [-0.5,0.12,-0.33],
          [-0.44,0.73,0.13]]

biases2=[-1,2,-0.5]

#Convert lists to numpy arrays
inputs_array=np.array(inputs)
weights_array=np.array(weights)
biases_array=np.array(biases)
weights2_array=np.array(weights2)
biases2_array=np.array(biases2)

#Calculate the output of the first layer

layer1_output=np.dot(inputs_array,weights_array.T)+biases_array
#print(layer1_output)

#Calculate the output of the second layer
layer2_output=np.dot(layer1_output,weights2_array.T)+biases2_array
print(layer2_output)
```
