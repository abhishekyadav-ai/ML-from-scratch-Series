```python
#coding first neuron
inputs=[1,2,3]
weights=[0.2,0.8,-0.5]
bias=2

outputs=(inputs[0]*weights[0] + inputs[1]*weights[1] + inputs[2]*weights[2] +bias)
print(outputs)

#coding the second neuron
inputs=[1.0,2.0,3.0,2.5]
weights=[0.2,0.8,-0.5,1.0]
bias=2.0
outputs=(inputs[0]*weights[0] + inputs[1]*weights[1] + inputs[2]*weights[2] +  inputs[3]*weights[3] +bias)
print(outputs)
```

```python
#coding the layers of neurons

inputs=[1,2,3,2.5]
weights=[[0.2,0.8,-0.5,1],
         [0.5,-0.91,0.26,-0.5],
         [-0.26,-0.27,0.17,0.87]]

bias=[2,3,0.5]
outputs=[
    #Neuron1:
    inputs[0]*weights[0][0]+
    inputs[1]*weights[0][1]+
    inputs[2]*weights[0][2]+
    inputs[3]*weights[0][3]+ bias[0],
    #Neuron2:
    inputs[0]*weights[1][0]+
    inputs[1]*weights[1][1]+
    inputs[2]*weights[1][2]+
    inputs[3]*weights[1][3]+ bias[1],
    #Neuron 3:
    inputs[0]*weights[2][0]+
    inputs[1]*weights[2][1]+
    inputs[2]*weights[2][2]+
    inputs[3]*weights[2][3]+ bias[2]

]
print(outputs)
```

What if there are 50 neurons we can not write the summation for 50 neurons right, so lets use loops

```python
inputs=[1.1,2.2,3.3]

#List of weights

weights=[[2,0,0],
         [2,2,0],
         [2,2,2]]
#list of biases
biases=[0,0,1]

#Output of current layer
layer_outputs=[]

#for each neuron
for neuron_weights, neuron_bias in zip(weights, biases):
    neuron_output=0
    for n_input, weight in zip(inputs, neuron_weights):
    #Multipyling this input with the associated weights
        neuron_output+=n_input*weight
    #Adding bias Term
    neuron_output+=neuron_bias
    #Storing the neuron outputs
    layer_outputs.append(neuron_output)
print(layer_outputs)


```
